```python
# 1. 파일 업로드
from google.colab import files
uploaded = files.upload()

# 2. 라이브러리 불러오기
import pandas as pd

# 3. 업로드한 파일명 자동 확인
file_name = next(iter(uploaded))

# 4. CSV 불러오기
df = pd.read_csv(
    file_name,
    encoding="utf-8-sig",
    dtype={
        "name": "string",
        "gacha_type": "int16",
        "item_type": "int8",
        "rank_type": "int8",
        "gacha_id": "string"
    },
    parse_dates=["gacha_time"]
)

# 5. 결과 확인
print("데이터 불러오기 완료")
print("파일명:", file_name)
print("전체 행 수:", len(df))
print("전체 열 수:", len(df.columns))
print("데이터 크기:", df.shape)

display(df.head(10))

print("\n배너별 기록 수")
display(df["gacha_type"].value_counts().sort_index())

print("\n등급별 기록 수")
display(df["rank_type"].value_counts().sort_index())

print("\n아이템 유형별 기록 수")
display(df["item_type"].value_counts().sort_index())

print("\n데이터 정보")
df.info()
```
